```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_excel("C:/Users/Lenovo/nephro_platform/Backend/data/kidney_disease.xlsx")
print(df[df['diagnostic_nom'] == "Glomérulonéphrite"].shape)
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y="diagnostic_nom", order=df['diagnostic_nom'].value_counts().index)
plt.title("Répartition des diagnostics dans le dataset")
plt.xlabel("Nombre de cas")
plt.ylabel("Diagnostic")
plt.show()
```
